```python
class Point:

    def __init__(self, x, y):
        self.x_cod = x
        self.y_cod = y

    def __str__(self):
        return f"<{self.x_cod},{self.y_cod}>"

    def euclidean_distance(self, other):
        return ((self.x_cod - other.x_cod)**2 +
                (self.y_cod - other.y_cod)**2) ** 0.5

    def distance_from_origin(self):
        return (self.x_cod**2 + self.y_cod**2) ** 0.5

    def mid_point(self, other):
        return f"<{(self.x_cod + other.x_cod)/2},{(self.y_cod + other.y_cod)/2}>"

    def internal_sec_point(self, other, m, n):
        return f"<{(m*other.x_cod + n*self.x_cod)/(m+n)}," \
               f"{(m*other.y_cod + n*self.y_cod)/(m+n)}>"

    def external_sec_point(self, other, m, n):
        return f"<{(m*other.x_cod - n*self.x_cod)/(m-n)}," \
               f"{(m*other.y_cod - n*self.y_cod)/(m-n)}>"

    def line_eq_two_point(self, other):
        if self.x_cod == other.x_cod:
            return f"x - {self.x_cod} = 0"

        m = (other.y_cod - self.y_cod) / (other.x_cod - self.x_cod)
        c = self.y_cod - m * self.x_cod

        return f"{m}x - y + {c} = 0"
```

```python
class Line:

    def __init__(self, A, B, C):
        self.A = A
        self.B = B
        self.C = C

    def __str__(self):
        return f"{self.A}x + {self.B}y + {self.C} = 0"

    def point_on_line(self, point):
        if self.A*point.x_cod + self.B*point.y_cod + self.C == 0:
            return "lies on the line"
        return "does not lie on the line"

    def shortest_distance(self, point):
        return abs(self.A*point.x_cod +
                   self.B*point.y_cod +
                   self.C) / ((self.A**2 + self.B**2) ** 0.5)
```

```python
class Circle:

  def __init__(self, g, f, c):
    self.g = g
    self.f = f
    self.c = c

  def __str__(self):
    return f"x^2 + y^2 + {2*self.g}x + {2*self.f}y + {self.c} = 0"

  def centre(self):
    return "<{},{}>".format(-self.g, -self.f)

  def radius(self):
    return (self.g**2 + self.f**2 - self.c)**0.5

  def point_on_circle(self, point):
    val = point.x_cod**2 + point.y_cod**2 + \
          2*self.g*point.x_cod + 2*self.f*point.y_cod + self.c
    if val == 0:
      return "lies on the circle"
    else:
      return "does not lie on the circle"

  def tangent(self, point):
    return "x*{} + y*{} + {}(x+{}) + {}(y+{}) + {} = 0".format(
        point.x_cod, point.y_cod,
        self.g, point.x_cod,
        self.f, point.y_cod,
        self.c
    )

  def normal(self, point):
    return "(x - {})/({}) = (y - {})/({})".format(
        point.x_cod, point.x_cod + self.g,
        point.y_cod, point.y_cod + self.f
    )
```

```python
class Parabola:

  def __init__(self, a):
    self.a = a

  def __str__(self):
    return "y^2 = {}x".format(4*self.a)

  def focus(self):
    return "<{},0>".format(self.a)

  def directrix(self):
    return "x = {}".format(-self.a)

  def point_on_parabola(self, point):
    if point.y_cod**2 == 4*self.a*point.x_cod:
      return "lies on the parabola"
    else:
      return "does not lie on the parabola"

  def tangent(self, point):
    return "y*{} = {}(x + {})".format(
        point.y_cod, 2*self.a, point.x_cod
    )

  def normal(self, point):
    m = -point.y_cod / (2*self.a)
    return "y - {} = {}(x - {})".format(
        point.y_cod, m, point.x_cod
    )
```

```python
class Ellipse:

  def __init__(self, a, b):
    self.a = a
    self.b = b

  def __str__(self):
    return "x^2/{} + y^2/{} = 1".format(self.a**2, self.b**2)

  def eccentricity(self):
    return ((self.a**2 - self.b**2)**0.5) / self.a

  def focus(self):
    c = (self.a**2 - self.b**2)**0.5
    return "<{},0> , <{},0>".format(c, -c)

  def latus_rectum(self):
    return (2*self.b**2) / self.a

  def point_on_ellipse(self, point):
    val = (point.x_cod**2)/(self.a**2) + (point.y_cod**2)/(self.b**2)
    if val == 1:
      return "lies on the ellipse"
    else:
      return "does not lie on the ellipse"

  def tangent(self, point):
    return "x*{}/{} + y*{}/{} = 1".format(
        point.x_cod, self.a**2,
        point.y_cod, self.b**2
    )

  def normal(self, point):
    return "(x - {})/({}) = (y - {})/({})".format(
        point.x_cod, point.x_cod*self.b**2,
        point.y_cod, point.y_cod*self.a**2
    )
```

```python
class Hyperbola:

  def __init__(self, a, b):
    self.a = a
    self.b = b

  def __str__(self):
    return "x^2/{} - y^2/{} = 1".format(self.a**2, self.b**2)

  def eccentricity(self):
    return ((self.a**2 + self.b**2)**0.5) / self.a

  def asymptotes(self):
    return "y = ±({}/{})x".format(self.b, self.a)

  def point_on_hyperbola(self, point):
    val = (point.x_cod**2)/(self.a**2) - (point.y_cod**2)/(self.b**2)
    if val == 1:
      return "lies on the hyperbola"
    else:
      return "does not lie on the hyperbola"

  def tangent(self, point):
    return "x*{}/{} - y*{}/{} = 1".format(
        point.x_cod, self.a**2,
        point.y_cod, self.b**2
    )

  def normal(self, point):
    return "(x - {})/({}) = (y - {})/(-{})".format(
        point.x_cod, point.x_cod*self.b**2,
        point.y_cod, point.y_cod*self.a**2
    )
```
